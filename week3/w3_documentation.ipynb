# Components
## 1.1 What are the inputs?
	•	Quiz 1 mark
	•	Quiz 2 mark
	•	Quiz 3 mark
	•	Choice (Y/N)

## 1.2 What is the process?
	•	Read the three quiz marks.
	•	Calculate the average.
	•	Check if the average is 50 or above.

## 1.3 What is the output?
	•	Average mark.
	•	Pass or Fail.
	•	“Program Ended” when the user exits.
